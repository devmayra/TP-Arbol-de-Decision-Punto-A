# Construcción de un Árbol de Decisión
## Predicción de Aceptación de Oferta de Plan de Datos Móviles

Una empresa de telecomunicaciones quiere predecir si un cliente aceptará una oferta de plan de datos móviles. Para ello, se dispone de un conjunto de datos con información de 10 clientes.

## Datos del conjunto

| ID | Edad | Uso de datos | Tiene línea fija | Aceptó oferta |
|:--:|:----:|:------------:|:----------------:|:-------------:|
| 1  | 24   | 2.5          | No               | No            |
| 2  | 38   | 6.0          | Sí               | Sí            |
| 3  | 29   | 3.0          | No               | No            |
| 4  | 45   | 8.0          | Sí               | Sí            |
| 5  | 52   | 7.5          | Sí               | Sí            |
| 6  | 33   | 4.0          | No               | No            |
| 7  | 41   | 5.5          | Sí               | Sí            |
| 8  | 27   | 2.0          | No               | No            |
| 9  | 36   | 6.5          | Sí               | Sí            |
| 10 | 31   | 3.5          | No               | No            |

## Fórmulas utilizadas

**Fórmula de Entropía:**
H = -(P(Sí) × log₂(P(Sí)) + P(No) × log₂(P(No)))

**Fórmula de Entropía Ponderada:**
H_ponderada = (Peso₁ × H₁) + (Peso₂ × H₂) + (Peso₃ × H₃) + ...

**Fórmula de Ganancia de Información:**
Ganancia = H_original - H_ponderada

**Cálculo de log₂:**
log₂(x) = log(x) / log(2)

---
# OBJETIVO 1: Calcular la entropía del conjunto original

## Análisis del conjunto completo

- Total de clientes: 10
- Clientes que aceptaron (Sí): 5
- Clientes que no aceptaron (No): 5

## Cálculo

P(Sí) = 5/10 = 0.5
P(No) = 5/10 = 0.5

log₂(0.5) = log(0.5) / log(2) = -0.301 / 0.301 = -1

H = -(0.5 × (-1) + 0.5 × (-1))
H = -(-0.5 - 0.5)
H = -(-1)

**H_original = 1.0**

---
# OBJETIVO 2: Evaluar la ganancia de información para los atributos

## ATRIBUTO 1: EDAD
Rangos: Joven (≤30), Adulto (31–50), Mayor (>50)

### Entropía de cada subconjunto de Edad

#### Subconjunto Joven (≤30): 3 clientes
Clientes: IDs 1, 3, 8

- Aceptó No: 3

P(Sí) = 0/3 = 0
P(No) = 3/3 = 1

**H_Joven = 0**

---

#### Subconjunto Adulto (31-50): 6 clientes
Clientes: IDs 2, 4, 6, 7, 9, 10
- Aceptó Sí: 4
- Aceptó No: 2

P(Sí) = 4/6 = 0.667
P(No) = 2/6 = 0.333

log₂(0.667) = log(0.667) / log(2) = -0.176 / 0.301 ≈ -0.585
log₂(0.333) = log(0.333) / log(2) = -0.477 / 0.301 ≈ -1.585

H = -(0.667 × (-0.585) + 0.333 × (-1.585))
H = -(-0.389 - 0.528)
H = -(-0.917)

**H_Adulto = 0.917**

---

#### Subconjunto Mayor (>50): 1 cliente
Clientes: ID 5
- Aceptó Sí: 1
- Aceptó No: 0

P(Sí) = 1/1 = 1
P(No) = 0/1 = 0

H = -(1 × log₂(1) + 0 × log₂(0))
H = -(0 + 0)

**H_Mayor = 0**

### Entropía ponderada de Edad

Peso_Joven = 3/10 = 0.3
Peso_Adulto = 6/10 = 0.6
Peso_Mayor = 1/10 = 0.1

H_ponderada = (0.3 × 0) + (0.6 × 0.917) + (0.1 × 0)
H_ponderada = 0 + 0.5502 + 0

**H_ponderada (Edad) = 0.5502**

### Ganancia de información - Edad

Ganancia = H_original - H_ponderada
Ganancia = 1.0 - 0.5502

**Ganancia (Edad) = 0.450**

# OBJETIVO 3: Construir el árbol de decisión paso a paso

## Nivel 0 (Raíz): División por Línea fija

Como **Línea fija** tiene la ganancia más alta (1.0), es el atributo que se elige para la raíz del árbol.

**División:**

- **Rama izquierda (Línea fija = Sí):** 5 clientes
  - 5 aceptaron la oferta
  - Hoja pura → Predicción: **SÍ**
  - Entropía: 0

- **Rama derecha (Línea fija = No):** 5 clientes
  - 0 aceptaron la oferta (5 rechazaron)
  - Hoja pura → Predicción: **NO**
  - Entropía: 0

**Conclusión:** El árbol es perfecto con solo un nodo raíz. Ambas ramas son puras (entropía = 0), por lo que no hay necesidad de divisiones adicionales. El árbol tiene profundidad 1.

## Representación del árbol de decisión

```
                    ¿TIENE LÍNEA FIJA?
                           |
                    /              \
                  SÍ                NO
                 /                    \
        ACEPTÓ = SÍ              ACEPTÓ = NO
        (5 clientes)             (5 clientes)
        Predicción: SÍ           Predicción: NO
        Entropía: 0              Entropía: 0
```

---
# OBJETIVO 4: Conclusión

## Mejor atributo para comenzar el árbol

**Atributo elegido:** `Tiene línea fija`

**Razón:**
- Ganancia de información: **1.0** (máxima ganancia posible)
- Reduce la entropía de 1.0 (máxima incertidumbre) a 0 (certeza total)
- Separa perfectamente el conjunto en dos grupos homogéneos
- Proporciona una clasificación 100% precisa

---

## Regla de predicción para la oferta

Para predecir si un cliente **aceptará la oferta de plan de datos móviles**, se utiliza la siguiente regla:

```
SI cliente tiene línea fija = Sí
  → Predicción: ACEPTARÁ LA OFERTA (Sí)

SI cliente tiene línea fija = No
  → Predicción: NO ACEPTARÁ LA OFERTA (No)
```

---

## Características del árbol de decisión

- **Profundidad:** 1 nivel (solo nodo raíz)
- **Número de nodos:** 3 (1 raíz + 2 hojas)
- **Nodos terminales:** 2 hojas puras
- **Precisión:** 100% (clasifica correctamente todos los casos)
- **Interpretabilidad:** Muy alta (regla extremadamente simple)
- **Complejidad:** Mínima

---

## Conclusión general

El atributo **"Tiene línea fija"** es un predictor extraordinariamente útil para determinar si un cliente aceptará la oferta de plan de datos móviles.

**Observación importante:** Existe una correlación perfecta entre tener línea fija y aceptar la oferta:
- **Todos los clientes con línea fija aceptaron la oferta**
- **Todos los clientes sin línea fija rechazaron la oferta**

Esto sugiere que la línea fija es un fuerte indicador del interés del cliente en servicios de telecomunicaciones adicionales.
